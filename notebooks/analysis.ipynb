```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/clean/clean_cafe_sales.csv", parse_dates=["Transaction Date"])


```

```python
known = df[df["Item"] != "Unknown"]
most_sold_items_known = known.groupby("Item")["Quantity"].sum().sort_values(ascending=False)
most_sold_items_known.plot(kind="bar", figsize=(10, 6), color="skyblue")
plt.title("Most Sold Items")
plt.xlabel("Item")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../data/clean/most_sold_items_known.png")
plt.show()
```

```python
item_sales = df.groupby("Item")["Total Spent"].sum().sort_values(ascending=False)
item_sales.plot(kind="bar", figsize=(10, 6), color="lightgreen")
plt.title("Total Sales by Item")
plt.xlabel("Item")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../data/clean/total_sales_by_item.png")
plt.show()

```

```python
monthly_transactions = df.groupby(known["Transaction Date"].dt.to_period("M"))["Transaction ID"].count()
monthly_transactions.plot(kind="line", figsize=(10, 6), marker='o', color='lightcoral')
plt.title("Monthly Transaction Trends")
plt.xlabel("Month")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../data/clean/monthly_transaction_trends.png")
plt.show()
```

```python
payment_method_sales = df.groupby("Payment Method")["Total Spent"].sum().sort_values(ascending=False)
payment_method_sales.plot(kind="bar", figsize=(10, 6), color="lightblue")
plt.title("Total Sales by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("../data/clean/total_sales_by_payment_method.png")
plt.show()
```

```python
location_sales = df.groupby("Location")["Total Spent"].sum().sort_values(ascending=False)
location_sales.plot(kind="bar", figsize=(10, 6), color="grey")
plt.title("Total Sales by Location")
plt.xlabel("Location")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("../data/clean/total_sales_by_location.png")
plt.show()
```
